## Question 15 – if-elif-else and Nested if

Ask the user to enter marks out of 100. Use `if-elif-else` to print `Excellent` for 80 or above, `Good` for 60–79, `Pass` for 40–59, and `Fail` below 40. First check that marks are between 0 and 100. For valid marks, use a nested `if` to display `Passed the examination` when marks are at least 40.

```python
# Write your answer below

marks = float(input("Enter marks out of 100: "))

if 0 <= marks <= 100:
    if marks >= 80:
        print("Excellent")
    elif marks >= 60:
        print("Good")
    elif marks >= 40:
        print("Pass")
    else:
        print("Fail")

    if marks >= 40:
        print("Passed the examination")
else:
    print("Invalid marks! Please enter a value between 0 and 100.")

```
